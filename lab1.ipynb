```python
def is_palindrome(value: str) -> bool:
    letters_only = ""
    for character in value:
        if character.isalpha():
            letters_only += character.lower()
    return letters_only == letters_only[::-1]

examples = [
    "А роза упала на лапу Азора",
    "Hello",
    "Was it a car or a cat I saw?",
]

for example in examples:
    print(example, "->", is_palindrome(example))
```

```python
def caesar(text: str, shift: int) -> str:
    encrypted = ""
    for character in text:
        if "A" <= character <= "Z":
            code = (ord(character) - ord("A") + shift) % 26
            encrypted += chr(ord("A") + code)
        elif "a" <= character <= "z":
            code = (ord(character) - ord("a") + shift) % 26
            encrypted += chr(ord("a") + code)
        else:
            encrypted += character
    return encrypted
message = "Este, es el mundo real"
shift = 10
encrypted_message = caesar(message, shift)
decrypted_message = caesar(encrypted_message, -shift)
print("Original:", message)
print("Cifrado:", encrypted_message)
print("Descifrado:", decrypted_message)
```

```python
def strange_case(text: str) -> str:
    letter_index = 0
    result = ""
    for character in text:
        if character.isalpha():
            if letter_index % 2 == 0:
                result += character.upper()
            else:
                result += character.lower()
            letter_index += 1
        else:
            result += character
    return result


print(strange_case("hello world"))
```

```python
from pathlib import Path
from urllib.request import urlopen


def download_text(url: str, path: str) -> None:
    with urlopen(url) as response:
        data = response.read()
    content = data.decode("utf-8")
    Path(path).write_text(content, encoding="utf-8")


def read_text(path: str) -> str:
    return Path(path).read_text(encoding="utf-8")


def clean_words(text: str) -> list[str]:
    punctuation = ".,!?;:\"'()[]"

    result = []
    for word in text.split():
        clean_word = word.strip(punctuation).lower()
        if clean_word:
            result.append(clean_word)
    return result


def find_capitalized(text: str) -> list[str]:
    result = []

    for word in text.split():
        if word and word[0].isupper():
            result.append(word)
    return result


def is_palindrome(text: str) -> bool:
    normalized = "".join(
        character.lower()
        for character in text
        if character.isalpha()
    )

    return normalized == normalized[::-1]


def caesar(text: str, shift: int) -> str:
    """Aplica un cifrado César a letras latinas."""

    result = ""

    for character in text:
        if "A" <= character <= "Z":
            base = ord("A")
        elif "a" <= character <= "z":
            base = ord("a")
        else:
            result += character
            continue

        shifted = (ord(character) - base + shift) % 26

        result += chr(base + shifted)

    return result


def char_stats(text: str) -> None:

    letters = digits = spaces = others = 0

    for character in text:
        if character.isalpha():
            letters += 1
        elif character.isdigit():
            digits += 1
        elif character.isspace():
            spaces += 1
        else:
            others += 1

    print(f"Letras: {letters}")
    print(f"Dígitos: {digits}")
    print(f"Espacios: {spaces}")
    print(f"Otros: {others}")


def pig_latin(word: str) -> str:

    vowels = "aeiouAEIOU"

    if not word:
        return word

    if word[0] in vowels:
        return word + "way"

    return word[1:] + word[0] + "ay"


def parse_url(url: str) -> tuple[str, str, str, str]:

    protocol, remainder = url.split("://", 1)
    parts = remainder.split("/")
    domain = parts[0]
    filename = parts[-1]
    extension = filename[filename.rfind("."):]
    return protocol, domain, filename, extension


def main() -> None:

    url = "https://www.gutenberg.org/files/74/74-0.txt"
    path = "article.txt"
    download_text(url, path)
    text = read_text(path)
    words = clean_words(text)
    print("Caracteres:", len(text))
    print("Líneas:", text.count("\n") + 1)
    print("Palabras:", len(words))
    char_stats(text)
    print("Capitalizadas:", find_capitalized(text)[:20])
    print(is_palindrome("А роза упала на лапу Азора"))
    print(caesar("Hello, World!", 3))
    print([pig_latin(word) for word in words[:20]])
    print(parse_url(url))
if __name__ == "__main__":
    main()
```
