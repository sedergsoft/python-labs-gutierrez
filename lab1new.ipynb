````python
import requests

def download_text_article(url: str, output_file: str = 'articulo.txt', timeout: int = 15) -> str:
    """
    Descarga un artículo/archivo de texto plano desde una URL y lo guarda localmente.
    """
    headers = {'User-Agent': 'Mozilla/5.0'}
    response = requests.get(url, headers=headers, timeout=timeout)
    response.raise_for_status()

    # Asegurar decodificación correcta del texto
    response.encoding = response.apparent_encoding or 'utf-8'

    with open(output_file, 'w', encoding='utf-8') as f:
        f.write(response.text)

    return output_file



````

Задание 3. Поиск имён через регулярные выражения.

Создайте функцию, которая принимает строку в качестве параметра и использует регулярные выражения для поиска всех имён (предположим, что имя — это слово с заглавной буквы).

````python
import re

def find_names(text: str) -> list:
    """Busca palabras que comienzan con mayúscula (nombres)."""
    pattern = r'\b[A-ZА-ЯЁ][a-zа-яё]*\b'
    return re.findall(pattern, text)


````

Задание 5. Подсчёт упоминаний имён.

````python
from collections import Counter

def count_name_mentions(filename: str) -> dict:
    """
    Tarea 5: recibe el nombre de un archivo con nombres (uno por línea)
    y devuelve un diccionario {nombre: cantidad de menciones}.
    """
    with open(filename, 'r', encoding='utf-8') as f:
        nombres = [linea.strip() for linea in f if linea.strip()]

    return dict(Counter(nombres))


````

Задание 7. Самое частое имя.

Создайте функцию, которая принимает словарь (созданный на предыдущем шаге) и выводит имя, которое упоминается чаще всего, и количество его упоминаний.

````python
def most_frequent_name(names_count: dict) -> tuple:
    """
    Recibe un diccionario {nombre: cantidad} y devuelve
    el nombre más frecuente y su cantidad de menciones.
    """
    if not names_count:
        return None, 0

    nombre_mas_frecuente = max(names_count, key=names_count.get)
    cantidad = names_count[nombre_mas_frecuente]
    return nombre_mas_frecuente, cantidad

````

## 1. Стандартные типы данных в Python. Отличие list от tuple

**Стандартные типы данных:** `int`, `float`, `complex`, `str`, `bool`, `list`, `tuple`, `set`, `frozenset`, `dict`, `bytes`, `bytearray`, `NoneType`.

**Отличие `list` от `tuple`:**
- `list` — изменяемый (можно добавлять, удалять и изменять элементы), записывается в квадратных скобках: `[1, 2, 3]`.
- `tuple` — неизменяемый (после создания изменить нельзя), записывается в круглых скобках: `(1, 2, 3)`.

## 2. Глобальные и локальные переменные

- **Глобальные переменные** объявляются вне функций и доступны во всей программе.
- **Локальные переменные** объявляются внутри функции и существуют только во время её выполнения.

Чтобы изменить глобальную переменную внутри функции, используют ключевое слово `global`:

```python
x = 10

def func():
    global x
    x += 1
```

## 3. Разница между return и yield

- **`return`** завершает функцию и возвращает одно значение.
- **`yield`** приостанавливает функцию и возвращает значение по одному, превращая её в генератор; при следующем вызове выполнение продолжается с места остановки.

## 4. Операторы break и continue

- **`break`** полностью прерывает выполнение цикла.
- **`continue`** пропускает оставшуюся часть текущей итерации и переходит к следующей.

## 5. Регулярные выражения и модуль re

**Регулярное выражение** — это шаблон для поиска, извлечения или замены текста по определённому правилу.

**Функции модуля `re`:** `re.match()`, `re.search()`, `re.findall()`, `re.finditer()`, `re.sub()`, `re.split()`, `re.compile()`.

## 6. Отличие re.search() от re.findall()

- **`re.search()`** возвращает только первое совпадение в виде объекта (или `None`, если совпадений нет).
- **`re.findall()`** возвращает список всех совпадений (или пустой список, если совпадений нет).

## 7. Модуль и библиотека

- **Модуль** — отдельный файл с расширением `.py`, содержащий функции, классы и переменные. Примеры: `math`, `os`.
- **Библиотека** — набор модулей с общей функциональностью. Примеры: `requests`, `numpy`, `BeautifulSoup`.

## 8. Функции модуля os и os.path.join()

**Функции модуля `os`:** `os.getcwd()`, `os.listdir()`, `os.mkdir()`, `os.remove()`, `os.rename()`, `os.path.exists()`, `os.path.join()`.

**`os.path.join()`** используется для безопасного объединения частей пути с учётом разделителя конкретной операционной системы (`/` или `\`).

## 9. Правила оформления модулей

- Имя файла — строчными буквами с подчёркиваниями: `my_module.py`.
- В начале файла — докстринг с описанием.
- Импорты — в начале файла.
- Исполняемый код — под защитой `if __name__ == '__main__':`.
- Константы — заглавными буквами.
- Один модуль — одна чёткая задача.
